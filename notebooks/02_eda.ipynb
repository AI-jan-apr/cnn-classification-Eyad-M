# 02 — Exploratory Data Analysis

Explores the Dogs vs. Cats dataset: class distribution, image sizes, and sample visualisations.

## Imports

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from PIL import Image

try:
    import google.colab
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

if IN_COLAB:
    from google.colab import drive
    drive.mount('/content/drive')

DATA_DIR = '/content/drive/MyDrive/datasets/catsvsdogs' if IN_COLAB else '../data/raw'
splits_dir = Path(DATA_DIR) / 'splits'
train_df = pd.read_csv(splits_dir / 'train.csv')
```

## Class Distribution

```python
counts = train_df['class'].value_counts()
sns.barplot(x=counts.index, y=counts.values, palette='pastel')
plt.title('Class Distribution (Train Set)')
plt.ylabel('Count')
plt.show()
```

## Image Size Distribution

```python
sample = train_df.sample(300, random_state=42)
sizes = [Image.open(p).size for p in sample['path']]
widths, heights = zip(*sizes)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(widths, bins=30, color='steelblue')
axes[0].set_title('Width Distribution')
axes[1].hist(heights, bins=30, color='salmon')
axes[1].set_title('Height Distribution')
plt.tight_layout()
plt.show()
```

## Sample Images (Cats & Dogs)

```python
fig, axes = plt.subplots(2, 5, figsize=(15, 6))

for i, (_, row) in enumerate(train_df[train_df['class'] == 'Cat'].sample(5, random_state=1).iterrows()):
    axes[0][i].imshow(Image.open(row['path']))
    axes[0][i].set_title('Cat')
    axes[0][i].axis('off')

for i, (_, row) in enumerate(train_df[train_df['class'] == 'Dog'].sample(5, random_state=1).iterrows()):
    axes[1][i].imshow(Image.open(row['path']))
    axes[1][i].set_title('Dog')
    axes[1][i].axis('off')

plt.tight_layout()
plt.show()
```
